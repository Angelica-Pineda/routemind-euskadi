```python
from pyspark.sql.functions import current_timestamp, input_file_name
from pyspark.sql.types import StructType, StructField, StringType

source_path = "abfss://landing@stroutemindeuskadidev.dfs.core.windows.net/municipality_codes/municipality_codes.json"
delta_path = "abfss://bronze@stroutemindeuskadidev.dfs.core.windows.net/delta_tables/municipality_codes/data"
deltaTable = "dbw_routemind_euskadi_dev.bronze.municipality_codes"



mun_code_raw = spark.read \
    .json(source_path)
```

```python
mun_code_raw.count()
```

```python
df_air_quality_hist_bronze = mun_code_raw \
    .withColumn("ingestion_timestamp", current_timestamp()) \
    .withColumn("source_file", input_file_name())
```

```python
df_air_quality_hist_bronze.write \
    .format("delta") \
    .option("path", delta_path) \
    .option("mergeSchema", "true") \
    .mode("append") \
    .saveAsTable(deltaTable)


print(f"APPEND completed on {deltaTable}. rows processed: {df_air_quality_hist_bronze.count()}")
```
