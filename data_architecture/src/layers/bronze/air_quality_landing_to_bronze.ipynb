```python
from pyspark.sql.functions import current_timestamp, input_file_name
import json

source_path = "abfss://landing@stroutemindeuskadidev.dfs.core.windows.net/weather/air_quality/stations/air_quality_stations.json"
delta_path = "abfss://bronze@stroutemindeuskadidev.dfs.core.windows.net/delta_tables/air_quality_stations/data"
deltaTable = "dbw_routemind_euskadi_dev.bronze.air_quality_stations"



df_qa_station_raw = spark.read \
    .option("multiline", "true") \
    .json(source_path)
```

```python
df_qa_station_bronze = df_qa_station_raw \
    .withColumn("ingestion_timestamp", current_timestamp()) \
    .withColumn("source_file", input_file_name())
```

```python
df_qa_station_bronze.write \
    .format("delta") \
    .option("path", delta_path) \
    .option("overwriteSchema", "true") \
    .mode("overwrite") \
    .saveAsTable(deltaTable)


print(f"OVERWRITE completed on {deltaTable}. rows processed: {df_qa_station_bronze.count()}")
```
